```python
import os

def print_directory_structure(startpath, level=0):
    for element in os.listdir(startpath):
        path = os.path.join(startpath, element)
        print("|   " * level + "+-- " + element)
        if os.path.isdir(path):
            print_directory_structure(path, level + 1)

current_directory = os.getcwd() + "/NBA1/output1"
print(f"Directory structure of: {current_directory}")
print_directory_structure(current_directory)
```
